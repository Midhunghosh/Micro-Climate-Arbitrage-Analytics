```python
t25 = df["Temperature_C"].quantile(0.25)
t75 = df["Temperature_C"].quantile(0.75)

def temperature_category(value):

    if value <= t25:
        return "Lower Range"

    elif value <= t75:
        return "Middle Range"

    else:
        return "Upper Range"


df["Temperature_Category"] = (
    df["Temperature_C"].apply(temperature_category)
)
```

```python
h25 = df["Humidity_Percent"].quantile(0.25)
h75 = df["Humidity_Percent"].quantile(0.75)

def humidity_category(value):

    if value <= h25:
        return "Lower Range"

    elif value <= h75:
        return "Middle Range"

    else:
        return "Upper Range"


df["Humidity_Category"] = (
    df["Humidity_Percent"].apply(humidity_category)
)
```

```python
df["Distance_Saved_KM"] = (
    df["Distance_Primary_KM"] -
    df["Distance_Secondary_KM"]
)
```

```python
def reroute_status(row):

    if pd.isna(row["Spoilage_Risk_Percent"]):
        return "Risk Data Missing"

    elif (
        row["Spoilage_Risk_Percent"] == "High"
        and row["Distance_Saved_KM"] > 0
    ):
        return "Reroute Candidate"

    elif row["Spoilage_Risk_Percent"] == "High":
        return "Urgent Review"

    else:
        return "Monitor"


df["Reroute_Status"] = df.apply(
    reroute_status,
    axis=1
)
```
